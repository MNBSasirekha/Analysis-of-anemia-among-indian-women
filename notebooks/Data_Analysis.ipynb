```python
import pandas as pd
import numpy as np

# Load the dataset created in Stage 5
df = pd.read_excel("Anemia_Data_With_Risk_Score.xlsx")

print("Stage 5 dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
# Generate descriptive statistics for numerical variables

numeric_df = df.select_dtypes(include=np.number)

descriptive_statistics = numeric_df.describe().T

display(descriptive_statistics)
```

```python
# Analyze the distribution of women's anemia prevalence

women_anemia = df[
    "Women_Anemia"
]

print("Mean anemia prevalence:", women_anemia.mean())
print("Median anemia prevalence:", women_anemia.median())
print("Minimum anemia prevalence:", women_anemia.min())
print("Maximum anemia prevalence:", women_anemia.max())
print("Standard deviation:", women_anemia.std())
```

```python
# Calculate state-wise average women's anemia prevalence

state_analysis = (
    df.groupby("State")[
        "Women_Anemia"
    ]
    .agg(["mean", "median", "min", "max"])
    .sort_values("mean", ascending=False)
)

display(state_analysis)
```

```python
# Highest and lowest anemia prevalence districts

highest_anemia = (
    df.sort_values(
        "Women_Anemia",
        ascending=False
    )
    [
        [
            "District",
            "State",
            "Women_Anemia"
        ]
    ]
    .head(10)
)

lowest_anemia = (
    df.sort_values(
        "Women_Anemia",
        ascending=True
    )
    [
        [
            "District",
            "State",
            "Women_Anemia"
        ]
    ]
    .head(10)
)

print("Top 10 districts with highest anemia prevalence:")
display(highest_anemia)

print("Top 10 districts with lowest anemia prevalence:")
display(lowest_anemia)
```

```python
# Calculate correlations between numerical health and socio-economic indicators

correlation_matrix = numeric_df.corr()

display(correlation_matrix)
```

```python
# Analyze how each numerical indicator is correlated with women's anemia

anemia_correlation = (
    correlation_matrix[
        "Women_Anemia"
    ]
    .sort_values(ascending=False)
)

display(anemia_correlation)
```

```python
# Analyze the Risk Score created in Stage 5

print("Risk Score statistics:")

display(
    df["Risk Score"].describe()
)
```

```python
# Examine the relationship between Risk Score and women's anemia prevalence

risk_correlation = df[
    [
        "Risk Score",
        "Women_Anemia"
    ]
].corr()

display(risk_correlation)
```
