```python
import pandas as pd
import numpy as np

df = pd.read_excel('/content/AnemiaPrevalence_Selected_cols.xlsx')
```

```python
missing = df.isnull().sum()

missing = missing[
    missing > 0
]

print(missing)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows =", duplicates)
```

```python
df = df.drop_duplicates()
```

```python
df.columns = df.columns.str.strip()
```

```python
print(df.dtypes)
```

```python
numeric_cols = df.columns[2:]

for col in numeric_cols:
    # Convert values such as "(13.6)" to "13.6"
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .str.replace(r'^\((.*)\)$', r'\1', regex=True)
    )

    df[col] = pd.to_numeric(
        df[col],
        errors='coerce'
    )

print("Numeric columns converted successfully.")
```

```python
# Rename columns for easier analysis

df.rename(columns={
    "District Names": "District",
    "State/UT": "State",

    "Women (age 15-49) who are literate4 (%)": "Women_Literacy",

    "Women (age 15-49)  with 10 or more years of schooling (%)": "Women_Schooling",

    "Women age 15-19 years who were already mothers or pregnant at the time of the survey (%)": "Teenage_Pregnancy",

    "Mothers who consumed iron folic acid for 100 days or more when they were pregnant (for last birth in the 5 years before the survey) (%)": "IFA_Consumption",

    "Institutional births (in the 5 years before the survey) (%)": "Institutional_Births",

    "Women (age 15-49 years) whose Body Mass Index (BMI) is below normal (BMI <18.5 kg/m2)21 (%)": "Low_BMI",

    "Population living in households that use an improved sanitation facility2 (%)": "Improved_Sanitation",

    "Households using clean fuel for cooking3 (%)": "Clean_Cooking_Fuel",

    "Households with any usual member covered under a health insurance/financing scheme (%)": "Health_Insurance",

    "Children age 6-59 months who are anaemic (<11.0 g/dl)22 (%)": "Children_Anemia",

    "Non-pregnant women age 15-49 years who are anaemic (<12.0 g/dl)22 (%)": "NonPregnant_Anemia",

    "Pregnant women age 15-49 years who are anaemic (<11.0 g/dl)22 (%)": "Pregnant_Anemia",

    "All women age 15-49 years who are anaemic22 (%)": "Women_Anemia",

    "All women age 15-19 years who are anaemic22 (%)": "Teen_Women_Anemia"
}, inplace=True)

print("Column names renamed successfully!")
```

```python
print(df.columns.tolist())
```

```python
target = 'Women_Anemia'

invalid = df[
    (df[target] < 0) |
    (df[target] > 100)
]

print("Invalid Records:")
display(invalid)
```

```python
# Fill remaining missing numerical values with the median
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

print("Missing numerical values filled using column medians.")
```

```python
print("Final Shape:", df.shape)

print("Missing Values:")
print(df.isnull().sum().sum())

print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
df.to_excel("AnemiaPrevalence_Selected_cols_Verified.xlsx",index = False);
```
