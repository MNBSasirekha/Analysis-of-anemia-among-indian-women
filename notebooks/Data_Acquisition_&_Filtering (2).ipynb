# 2. Data Acquisition and Filtering

**Purpose:** Keep only the columns needed for the district-level anemia study.

**Libraries used:** pandas and NumPy only.

Here, “filtering” means choosing relevant columns. We are not removing districts or states. Data cleaning is handled by a teammate in a separate notebook.

```python
import pandas as pd
import numpy as np
```

## Step 1: Load the original dataset

Use the file name `NFHS_DATASET.csv`. Keep it in the same folder as this notebook, or upload it to Google Colab with this exact name.

```python
file_name = "NFHS_DATASET.csv"
df = pd.read_csv(file_name)
print("Original data size:", df.shape)
```

## Step 2: Choose the columns needed for the project

We keep these columns for three simple reasons:

- **Location:** district and state names tell us where each result belongs.
- **Anemia results:** these columns show anemia percentages for women and children.
- **Possible related factors:** schooling, nutrition, pregnancy care, sanitation, clean cooking fuel, and health insurance may help us study differences in anemia levels.

The list below uses the column names from the original dataset.

```python
selected_columns = ['District Names', 'State/UT', 'Women (age 15-49) who are literate4 (%)', 'Women (age 15-49)  with 10 or more years of schooling (%)', 'Women age 15-19 years who were already mothers or pregnant at the time of the survey (%)', 'Mothers who consumed iron folic acid for 100 days or more when they were pregnant (for last birth in the 5 years before the survey) (%)', 'Institutional births (in the 5 years before the survey) (%)', 'Women (age 15-49 years) whose Body Mass Index (BMI) is below normal (BMI <18.5 kg/m2)21 (%)', 'Population living in households that use an improved sanitation facility2 (%)', 'Households using clean fuel for cooking3 (%)', 'Households with any usual member covered under a health insurance/financing scheme (%)', 'All women age 15-49 years who are anaemic22 (%)', 'Children age 6-59 months who are anaemic (<11.0 g/dl)22 (%)', 'Non-pregnant women age 15-49 years who are anaemic (<12.0 g/dl)22 (%)', 'Pregnant women age 15-49 years who are anaemic (<11.0 g/dl)22 (%)', 'All women age 15-19 years who are anaemic22 (%) ']
```

## Step 3: Check that the chosen columns are present

This checks for spelling differences in the column names before selecting them.

```python
missing_columns = [col for col in selected_columns if col not in df.columns]
if len(missing_columns) > 0:
    print("These columns were not found:")
    print(missing_columns)
else:
    print("All selected columns are available.")
```

## Step 4: Keep only the chosen columns

This makes a smaller table with the information needed for our project. It keeps all rows, so no districts or states are removed at this stage.

```python
if len(missing_columns) == 0:
    selected_df = df[selected_columns].copy()
    print("Selected data size:", selected_df.shape)
    selected_df.head()
```

## Step 5: Save the selected data

Save the smaller table as a Excel file so the next stage can use it. This step does not clean or change the values.

```python
if len(missing_columns) == 0:
    selected_df.to_excel("AnemiaPrevalence_Selected_cols.xlsx", index=False)
    print("Saved: AnemiaPrevalence_Selected_cols.csv")
```
