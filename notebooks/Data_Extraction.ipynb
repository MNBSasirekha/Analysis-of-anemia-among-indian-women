```python
import pandas as pd

# Load the dataset produced during Data Acquisition
df = pd.read_excel("AnemiaPrevalence_Selected_cols_Verified.xlsx")

print(df.shape)
display(df.head())
```

```python
location_cols = [
    'District',
    'State'
]
```

```python
anemia_cols = [
    "Women_Anemia",
    "Children_Anemia",
    "NonPregnant_Anemia",
    "Pregnant_Anemia",
    "Teen_Women_Anemia"
]
```

```python
factor_cols = [
    "Women_Literacy",
    "Women_Schooling",
    "Teenage_Pregnancy",
    "IFA_Consumption",
    "Institutional_Births",
    "Low_BMI",
    "Improved_Sanitation",
    "Clean_Cooking_Fuel",
    "Health_Insurance"
]
```

```python
location_df = df[location_cols]

anemia_df = df[location_cols + anemia_cols]

factors_df = df[location_cols + factor_cols]
```

```python
location_df.to_excel("Location_Data.xlsx", index=False)

anemia_df.to_excel("Anemia_Data.xlsx", index=False)

factors_df.to_excel("Anemia_Related_Factors.xlsx", index=False)
```
