```python
import pandas as pd
import numpy as np

# Load the cleaned dataset produced by Data Validation & Cleaning
df = pd.read_excel("AnemiaPrevalence_Selected_cols_Verified.xlsx")

print("Cleaned dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
state_anemia = (
    df.groupby("State")[
        "Women_Anemia"
    ]
    .mean()
    .sort_values(ascending=False)
)

print("State-wise average anemia prevalence:")
display(state_anemia)
```

```python
# Create a summarized representation of important indicators at state level

state_summary = (
    df.groupby("State")
    .agg(
        Average_Women_Anemia=(
            "Women_Anemia",
            "mean"
        ),
        Average_Children_Anemia=(
            "Children_Anemia",
            "mean"
        ),
        Average_Low_BMI=(
            "Low_BMI",
            "mean"
        ),
        Average_Literacy=(
            "Women_Literacy",
            "mean"
        ),
        Average_IFA_Consumption=(
            "IFA_Consumption",
            "mean"
        )
    )
    .reset_index()
)

display(state_summary.head(10))
```

```python
# Calculate district-level anemia prevalence

district_anemia = df[
    [
        "District",
        "State",
        "Women_Anemia"
    ]
].copy()

district_anemia = district_anemia.sort_values(
    by="Women_Anemia",
    ascending=False
)

display(district_anemia.head(10))
```

```python
# Identify the 10 districts with the highest women's anemia prevalence

top10_districts = district_anemia.head(10)

display(top10_districts)
```

```python
# Create a composite Risk Score for each district

df["Risk Score"] = (
    0.35 * df["Women_Anemia"]
    + 0.20 * df["Low_BMI"]
    + 0.15 * (100 - df["Improved_Sanitation"])
    + 0.15 * (100 - df["Women_Literacy"])
    + 0.15 * (100 - df["IFA_Consumption"])
)

print("Risk Score created successfully.")

display(
    df[
        ["District", "State", "Risk Score"]
    ].head()
)
```

```python
# Rank districts according to their Risk Score

risk_ranking = (
    df[
        [
            "District",
            "State",
            "Women_Anemia",
            "Risk Score"
        ]
    ]
    .sort_values(
        by="Risk Score",
        ascending=False
    )
    .reset_index(drop=True)
)

print("District risk ranking created successfully.")

display(risk_ranking.head(15))
```

```python
# Categorize districts according to their Risk Score

df["Risk Category"] = pd.cut(
    df["Risk Score"],
    bins=[-np.inf, 25, 50, 75, np.inf],
    labels=[
        "Low",
        "Moderate",
        "High",
        "Very High"
    ]
)

display(
    df[
        [
            "District",
            "State",
            "Risk Score",
            "Risk Category"
        ]
    ].head(10)
)
```

```python
# Count the number of districts in each Risk Category

risk_category_summary = (
    df["Risk Category"]
    .value_counts()
    .sort_index()
)

display(risk_category_summary)
```

```python
# Save the aggregated results for use in later stages

state_summary.to_excel(
    "State_Wise_Anemia_Summary.xlsx",
    index=False
)

risk_ranking.to_excel(
    "District_Risk_Ranking.xlsx",
    index=False
)

df.to_excel(
    "Anemia_Data_With_Risk_Score.xlsx",
    index=False
)

print("Stage 5 outputs saved successfully!")
```
