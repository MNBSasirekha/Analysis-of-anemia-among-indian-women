**Data Acquisition**

```python
import pandas as pd

df = pd.read_csv('/content/NFHS_DATASET.csv')

if(not df.empty):
  print('Dataframe Created')
print(df.columns)
```

```python
# Selecting Relevant Columns

anemia_df = df[
[
    'District Names',
    'State/UT',
    'Women (age 15-49) who are literate4 (%)',
    'Women (age 15-49)  with 10 or more years of schooling (%)',
    'Women age 15-19 years who were already mothers or pregnant at the time of the survey (%)',
    'Mothers who consumed iron folic acid for 100 days or more when they were pregnant (for last birth in the 5 years before the survey) (%)',
    'Institutional births (in the 5 years before the survey) (%)',
    'Women (age 15-49 years) whose Body Mass Index (BMI) is below normal (BMI <18.5 kg/m2)21 (%)',
    'Population living in households that use an improved sanitation facility2 (%)',
    'Households using clean fuel for cooking3 (%)',
    'Households with any usual member covered under a health insurance/financing scheme (%)',
    'All women age 15-49 years who are anaemic22 (%)',
    'Children age 6-59 months who are anaemic (<11.0 g/dl)22 (%)',
    'Non-pregnant women age 15-49 years who are anaemic (<12.0 g/dl)22 (%)',
    'Pregnant women age 15-49 years who are anaemic (<11.0 g/dl)22 (%)',
    'All women age 15-19 years who are anaemic22 (%) '
]]

print(anemia_df.shape)

display(anemia_df.head())

anemia_df.to_excel("AnemiaPrevalence_Selected_cols.xlsx", index=False)
```

```python
# Statical Summary of Selected columns
display(anemia_df.describe().T)
```

**Filtering**

```python
# Filter data for multiple specific states
target_states = ['West Bengal', 'Bihar', 'Assam']
states_df = anemia_df[anemia_df['State/UT'].isin(target_states)]

display(states_df[['District Names', 'State/UT', 'All women age 15-49 years who are anaemic22 (%)']])
```

```python
# Districts with high anemia AND low female literacy
filtered_df = anemia_df[
    (anemia_df['All women age 15-49 years who are anaemic22 (%)'] > 60) &
    (anemia_df['Women (age 15-49) who are literate4 (%)'] < 60)
]
display(filtered_df[['District Names', 'State/UT', 'All women age 15-49 years who are anaemic22 (%)', 'Women (age 15-49) who are literate4 (%)']])
```

```python
# Top 10 High anemic States
state_anemia = anemia_df.groupby('State/UT')['All women age 15-49 years who are anaemic22 (%)'].mean().reset_index()

high_anemia_states = state_anemia.sort_values(by='All women age 15-49 years who are anaemic22 (%)', ascending=False)

display(high_anemia_states.head(10))
```

```python
#Caluculation of Top 5% Anemic Districts
# Calculate the threshold for the 95th percentile of anemia
threshold = anemia_df['All women age 15-49 years who are anaemic22 (%)'].quantile(0.95)

# Filter districts above this threshold
top_5_percent = anemia_df[anemia_df['All women age 15-49 years who are anaemic22 (%)'] >= threshold].sort_values(by='All women age 15-49 years who are anaemic22 (%)', ascending=False)
display(top_5_percent[['District Names', 'State/UT', 'All women age 15-49 years who are anaemic22 (%)']])
```
